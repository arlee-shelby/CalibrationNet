```python
fall = list(range(8622, 8866))
sets = {
    "Fall 2025 UDET": calibration_summary(fall, "nabpy-standard", "upper"),
    "Fall 2025 LDET": calibration_summary(fall, "short-trap-Fall2025", "lower"),
    "9469 UDET":      calibration_summary([9469], "nabpy-standard", "upper"),
    "9469 LDET":      calibration_summary([9469], "nabpy-standard", "lower"),
}

# 1. gain ± error
# 2. offset ± error        -> df.constant_kev,     yerr=df.constant_error
# 3. CE 976 peak width     -> df.ce976_sigma_kev,  yerr=df.ce976_sigma_error_kev
#    (also available in ADC: df.ce976_sigma_adc / ce976_sigma_error_adc)
```

```python
nrows,ncols=1,1
py.figure(figsize=((8)*ncols,(6)*nrows))
ax=py.subplot(nrows,ncols,1)
for name, df in sets.items():
    if name=="9469 LDET" or name=="Fall 2025 LDET":
        ax.errorbar(df.pixel-1000, df.gain_kev_per_adc, yerr=df.gain_error,fmt="o", label=name,alpha=0.5)
    else:
        ax.errorbar(df.pixel, df.gain_kev_per_adc, yerr=df.gain_error,fmt="o", label=name,alpha=0.5)

ax.set_xlabel("Pixel",fontsize=20)
ax.set_ylabel("Gain (keV/ADC)",fontsize=20)
ax.legend(fontsize=15)
ax.set_title("Gain",fontsize=20)
ax.tick_params(axis='both', which='major', labelsize=20)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from calibrationnet.queries import spectrum, stored_fit_curve

# ---- what to compare (any pixels/segments with stored CE fits) ----
RUN = 9469
PIXELS = [
    # (pixel, segment, color, label)
    (40,   26, "tab:blue", "UDET pixel 40"),
    (1062,  3, "tab:red",  "LDET pixel 1062"),
]
ZOOM_BELOW, ZOOM_ABOVE = 60, 40    # ADC window around the peak

# fig, ax = plt.subplots(figsize=(9, 5.5))
nrows,ncols=1,1
py.figure(figsize=((8)*ncols,(6)*nrows))
ax=py.subplot(nrows,ncols,1)

for pix, seg, color, label in PIXELS:
    edges, counts = spectrum(RUN, pix, segment=seg)       # raw histogram
    x, y = stored_fit_curve(RUN, pix, segment=seg)        # stored fit model

    # locate the CE 976 peak: the model maximum in the 976 region
    sel = (x > 2700) & (x < 3150)
    peak_adc = x[sel][np.argmax(y[sel])]

    lo, hi = int(peak_adc - ZOOM_BELOW), int(peak_adc + ZOOM_ABOVE)
    ys = counts[lo:hi].astype(float)
    norm = ys.max()                                       # peak-normalize

    ax.stairs(ys / norm, np.arange(lo, hi + 1) - peak_adc,
              color=color, alpha=0.6)                     # data
    msel = (x >= lo) & (x < hi)
    ax.plot(x[msel] - peak_adc, y[msel] / norm,
            color=color, lw=2, label=label)               # fit curve

# ax.axhline(0.5, ls="--", color="grey", alpha=0.5)         # half-max guide
ax.set_xlabel("ADC Relative to 976 Peak Centroid",fontsize=20)
ax.set_ylabel("Normalized Counts",fontsize=20)
ax.set_title(f"Run {RUN}: 976 keV peak shape, UDET vs LDET",fontsize=20)
ax.legend(fontsize=15)
ax.tick_params(axis='both', which='major', labelsize=20)
plt.show()
```

```python
import pylab as py
import numpy as np
from calibrationnet.queries import calibration_map, peak_table

PIXELS = [(8651,0,72), (8842,0,77), (8693,0,99),            # 2025 UDET
          (8839,0,1105), (8690,0,1109), (8681,0,1048),      # 2025 LDET
          (9469,13,18), (9469,51,67), (9469,33,53),         # 2026 UDET
          (9469,44,1078), (9469,19,1061), (9415,6,1054)]    # 2026 LDET
runs = sorted({p[0] for p in PIXELS})
full = calibration_map(run_numbers=runs, label="simulation")
ce   = calibration_map(run_numbers=runs, label="simulation-ce", current_only=False)

def pick(df, run, seg, pix):
    return df[(df.run==run)&(df.segment==seg)&(df.pixel==pix)].iloc[0]

x = np.arange(len(PIXELS))
labels = [f"{r}\np{p if p<1000 else p-1000}{'L' if p>=1000 else ''}" for r,s,p in PIXELS]

nrows,ncols = 1,2
py.figure(figsize=((8)*ncols,(6)*nrows))
for panel, (col, err, ylab) in enumerate([
        ("gain_kev_per_adc", "gain_error", "Gain (keV/ADC)"),
        ("constant_kev", "constant_error", "Offset (keV)")], start=1):
    ax = py.subplot(nrows, ncols, panel)
    for name, df, fmt in [("all 8 peaks", full, "o"), ("CE only", ce, "s")]:
        vals = [pick(df, *pt) for pt in PIXELS]
        ax.errorbar(x, [v[col] for v in vals], yerr=[v[err] for v in vals],
                    fmt=fmt, label=name, alpha=0.5)
    ax.set_xticks(x); ax.set_xticklabels(labels, fontsize=10)
    ax.set_ylabel(ylab, fontsize=20)
    ax.set_title(ylab.split(" ")[0], fontsize=20)
    ax.legend(fontsize=15)
    ax.tick_params(axis="y", which="major", labelsize=20)
py.tight_layout(); py.show()

# all fitted peak centroids/uncertainties behind any pixel:
peak_table(9469, 53, segment=33)                                  # 2026
peak_table(8839, 1105, segment=0, tf_label="short-trap-Fall2025") # 2025 LDET
```
